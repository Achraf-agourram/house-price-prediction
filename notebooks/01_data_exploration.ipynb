```python
numeric_columns = df.select_dtypes(include="number").columns.tolist()
categorical_columns = df.select_dtypes(include=["object", "category", "bool"]).columns.tolist()

print("Numeric:", len(numeric_columns))
print(numeric_columns)
print("\nCategorical:", len(categorical_columns))
print(categorical_columns)
```

## Analyse de SalePrice

```python
saleprice = df["SalePrice"]

print("Mean:", saleprice.mean())
print("Median:", saleprice.median())
print("Minimum:", saleprice.min())
print("Maximum:", saleprice.max())
print("Standard deviation:", saleprice.std())
print("Skewness:", saleprice.skew())
```

## Vérification d'incohérences

```python
checks = {
    "SalePrice < 0": (df["SalePrice"] < 0).sum(),
    "LotArea <= 0": (df["LotArea"] <= 0).sum(),
    "GrLivArea <= 0": (df["GrLivArea"] <= 0).sum(),
    "OverallQual outside 1-10": (~df["OverallQual"].between(1, 10)).sum(),
    "OverallCond outside 1-10": (~df["OverallCond"].between(1, 10)).sum(),
    "MoSold outside 1-12": (~df["MoSold"].between(1, 12)).sum()
}

pd.Series(checks)
```

## Valeurs aberrantes avec IQR

```python
def count_iqr_outliers(series):
    series = series.dropna()
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    return ((series < lower) | (series > upper)).sum()

outliers = {}

for column in numeric_columns:
    if column == "Id":
        continue
    count = count_iqr_outliers(df[column])
    if count > 0:
        outliers[column] = count

pd.Series(outliers).sort_values(ascending=False).head(10)
```

## Première gestion des absences

Certaines valeurs manquantes signifient réellement qu'une caractéristique n'existe pas, par exemple pas de garage, pas de sous-sol, pas de piscine ou pas de clôture. Nous les représentons donc explicitement.


```python
absence_categorical = [
    "Alley", "MasVnrType", "BsmtQual", "BsmtCond",
    "BsmtExposure", "BsmtFinType1", "BsmtFinType2",
    "FireplaceQu", "GarageType", "GarageFinish",
    "GarageQual", "GarageCond", "PoolQC", "Fence",
    "MiscFeature"
]

absence_numeric = [
    "MasVnrArea", "BsmtFinSF1", "BsmtFinSF2",
    "BsmtUnfSF", "TotalBsmtSF", "BsmtFullBath",
    "BsmtHalfBath", "GarageCars", "GarageArea"
]

df_clean = df.copy()

for column in absence_categorical:
    df_clean[column] = df_clean[column].fillna("None")

for column in absence_numeric:
    df_clean[column] = df_clean[column].fillna(0)

remaining_missing = df_clean.isnull().sum()
remaining_missing[remaining_missing > 0].sort_values(ascending=False)
```
